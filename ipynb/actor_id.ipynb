```python
import pandas as pd
import numpy as np
import requests
```

```python
df = pd.read_csv("actors.csv" , encoding='utf8')
```

```python
TMDB_API_KEY = "fce0af3409e6113c9b3c75aaf49341bb"
```

```python
def get_actor_id(actor_name, index):
    print(f"Processing row {index + 1}: Searching for actor '{actor_name}'...")

    url = f"https://api.themoviedb.org/3/search/person?api_key={TMDB_API_KEY}&query={actor_name}"
    response = requests.get(url)
    
    if response.status_code == 200:
        data = response.json()
        if data["results"]:
            actor_id = data["results"][0]["id"]
            print(f"✅ Found Actor ID: {actor_id} for '{actor_name}'\n")
            return actor_id
        else:
            print(f"❌ No results found for '{actor_name}'\n")
            return None
    else:
        print(f"⚠️ Error: {response.status_code} while searching for '{actor_name}'\n")
        return None
```

```python
df["actor_id"] = [get_actor_id(actor, idx) for idx, actor in enumerate(df["actor_name"])]
```
